```python
import pandas as pd

def normalize_dataset(input_csv, output_csv):
    df = pd.read_csv(input_csv)
    print(f"Shape initiale : {df.shape}")

    # ── Étape 1 : supprimer les colonnes brutes ──────────────────────────────
    cols_to_drop = [
        "darija_arabic",
        "darija_arabizi", 
        "english",
        "modern_standard_arabic",
        # colonnes inutiles Label Studio
        "annotation_id", "annotator", "created_at", "updated_at",
        "lead_time", "decision", "final_comment",
    ]
    df = df.drop(columns=[c for c in cols_to_drop if c in df.columns])

    # ── Étape 2 : renommer les colonnes corrected → noms finaux ──────────────
    df = df.rename(columns={
        "corrected_darija_arabic":  "darija_arabic",
        "corrected_darija_arabizi": "darija_arabizi",
        "corrected_english":        "english",
        "corrected_msa":            "modern_standard_arabic",
    })

    # ── Étape 3 : forcer status VALIDATED ────────────────────────────────────
    df["status"] = "VALIDATED"

    # ── Étape 4 : recalculer english_word_count ───────────────────────────────
    df["english_word_count"] = df["english"].fillna("").apply(
        lambda x: len(str(x).split())
    )

    # ── Étape 5 : ordre final des colonnes ────────────────────────────────────
    final_columns = [
        "data_id", "id", "classe",
        "darija_arabic", "darija_arabizi",
        "english", "modern_standard_arabic",
        "english_word_count", "status",
    ]
    df = df[final_columns]

    # ── Vérification ──────────────────────────────────────────────────────────
    print(f"Shape finale       : {df.shape}")
    print(f"Colonnes           : {df.columns.tolist()}")
    print(f"Statuts uniques    : {df['status'].unique()}")
    print(f"Valeurs manquantes :\n{df.isnull().sum()}")
    print(df.head(3).to_string())

    df.to_csv(output_csv, index=False, encoding="utf-8-sig")
    print(f"\n✓ Exporté → {output_csv}")

normalize_dataset("./deliverables/gold_final.csv", "silver_normalized.csv")
```
